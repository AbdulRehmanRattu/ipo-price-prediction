# Import Necessary Libraries

```python
# Cell 1
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from scipy import stats

print("Libraries imported successfully.")
```

# Load Dataset and Initial Inspection

```python
# Cell 2
# This cell loads the dataset from the provided Excel file and displays basic information.

file_path = "FINAL.xlsx"
df = pd.read_excel(file_path, sheet_name='Sheet1')

# Display first few rows of the dataset
display(df.head())

# Display dataset information
df.info()

# Check for missing values
missing_values = df.isnull().sum()
print("\nMissing Values in Each Column:\n", missing_values[missing_values > 0])
```

# Data Cleaning and Preprocessing


```python
# Cell 3: Data Cleaning and Preprocessing
# This cell handles missing values, converts necessary columns to numeric types, drops unnecessary columns, 
# removes highly correlated features, and encodes categorical variables.

# Display columns before preprocessing
print("Columns before preprocessing:")
print(df.columns.tolist())

# Drop unnecessary columns except 'Name of the IPO' (required for sector analysis)
df.drop(columns=['S. No.'], inplace=True, errors='ignore')

# Dynamically identify numeric columns that actually exist in the dataset
numeric_columns = [
    'Offer Price', 'Issue Size (cr.)', 'D/E RATIO', 'Interest Rate (RBI Repo Rate)', 'Unemployment Rate (%)',
    'Promoter Holding', 'Total Subscription Rate', 'Number of Shares Offered', 'Sales (Rs. Million)', 
    'PAT (Rs. Million)', 'Inflation Rate', 'GDP Growth Rate', 'Stock Market Index Performance (Nifty 50, BSE Sensex)', 
    'Market Volatility (VIX Index)', 'Listing Delay', 'Age'
]
existing_numeric_columns = [col for col in numeric_columns if col in df.columns]  # Ensure columns exist before processing

# Convert relevant columns to numeric, handling potential formatting issues
for col in existing_numeric_columns:
    df[col] = df[col].astype(str).str.replace('%', '', regex=True).str.replace(',', '', regex=True)  # Remove % and commas
    df[col] = pd.to_numeric(df[col], errors='coerce')  # Convert to numeric

# Fill missing values with column median
df.fillna(df.median(numeric_only=True), inplace=True)

# Encode categorical variable: Industry Classification
label_encoder = LabelEncoder()
df['Industry Classification Encoded'] = label_encoder.fit_transform(df['Industry Classification'])

# Retain 'Industry Classification' column to map IPOs with their respective sectors during analysis
df.drop(columns=['Industry Classification'], inplace=True, errors='ignore')

# Identify and remove highly correlated features
correlation_threshold = 0.8
correlation_matrix = df.select_dtypes(include=[np.number]).corr().abs()
upper_triangle = correlation_matrix.where(np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool))
highly_correlated_columns = [column for column in upper_triangle.columns if any(upper_triangle[column] > correlation_threshold)]

# Ensure important macroeconomic variables are not dropped
important_features = ['Interest Rate (RBI Repo Rate)', 'Unemployment Rate (%)']
highly_correlated_columns = [col for col in highly_correlated_columns if col not in important_features or col in df.columns]

df.drop(columns=highly_correlated_columns, inplace=True, errors='ignore')

# Display updated dataset information
df.info()

# Display columns after preprocessing
print("\nColumns after preprocessing:")
print(df.columns.tolist())

# Verify no missing values remain
print("\nRemaining Missing Values:", df.isnull().sum().sum())
```

# Exploratory Data Analysis (EDA)

```python
# Cell 4: Exploratory Data Analysis (EDA)
# This cell performs visual analysis to understand IPO performance across different industries.

# Map encoded values back to original industry names for better readability in plots
df['Industry Label'] = label_encoder.inverse_transform(df['Industry Classification Encoded'])

# Set figure size for better visualization
plt.figure(figsize=(12, 6))

# Boxplot for IPO Underpricing across industries
sns.boxplot(x='Industry Label', y='Underpricing (%)', data=df)
plt.xticks(rotation=90)
plt.title("IPO Underpricing Across Industries")
plt.xlabel("Industry")
plt.ylabel("Underpricing (%)")
plt.show()

# Boxplot for One-Year Returns across industries
plt.figure(figsize=(12, 6))
sns.boxplot(x='Industry Label', y='One Year Return (%)', data=df)
plt.xticks(rotation=90)
plt.title("One Year Returns Across Industries")
plt.xlabel("Industry")
plt.ylabel("One Year Return (%)")
plt.show()

# Correlation Heatmap - Using Only Numeric Columns
numeric_df = df.select_dtypes(include=[np.number])  # Select only numeric columns

plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title("Correlation Heatmap of IPO Dataset (Numerical Features Only)")
plt.show()
```

# Machine Learning Model Preparation

```python
# Cell 5: Machine Learning Model Preparation
# This cell prepares the dataset for training machine learning models to predict IPO underpricing and one-year returns.

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Automatically fetch feature columns (excluding target variables and 'Name of the IPO')
target_columns = ['Underpricing (%)', 'One Year Return (%)', 'Name of the IPO', 'Industry Label']
features = [col for col in df.columns if col not in target_columns]

# Ensure all feature columns are numeric
X = df[features].copy()  # Work on a copy to avoid modifying the original dataframe
for col in X.columns:
    if X[col].dtype == 'object':  # Convert non-numeric columns
        X[col] = X[col].astype(str).str.replace('%', '', regex=True).str.replace(',', '', regex=True)  # Remove symbols
        X[col] = pd.to_numeric(X[col], errors='coerce')  # Convert to numeric

# Handle any remaining missing values
X.fillna(X.median(numeric_only=True), inplace=True)

# Define target variables
y_underpricing = df['Underpricing (%)']
y_return = df['One Year Return (%)']

# Splitting data into training and testing sets (without stratification due to small class sizes)
X_train_up, X_test_up, y_train_up, y_test_up = train_test_split(
    X, y_underpricing, test_size=0.2, random_state=42, shuffle=True
)
X_train_ret, X_test_ret, y_train_ret, y_test_ret = train_test_split(
    X, y_return, test_size=0.2, random_state=42, shuffle=True
)

# Standardizing only numerical features
scaler = StandardScaler()
X_train_up = scaler.fit_transform(X_train_up)
X_test_up = scaler.transform(X_test_up)
X_train_ret = scaler.fit_transform(X_train_ret)
X_test_ret = scaler.transform(X_test_ret)

# Display final selected features
print("Final features used for machine learning models:", features)
print("Data successfully prepared for machine learning models.")
```

# Model Training & Evaluation

```python
# Cell 6
# This cell trains machine learning models to predict IPO underpricing and one-year returns and evaluates their performance.

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Function to train and evaluate a model
def train_and_evaluate_model(model, X_train, X_test, y_train, y_test, target_name):
    # Train the model
    model.fit(X_train, y_train)
    
    # Make predictions
    y_pred = model.predict(X_test)
    
    # Evaluate the model
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = mse ** 0.5
    r2 = r2_score(y_test, y_pred)

    # Display results
    print(f"\nPerformance for {target_name} using {model.__class__.__name__}:")
    print(f"Mean Absolute Error (MAE): {mae:.4f}")
    print(f"Mean Squared Error (MSE): {mse:.4f}")
    print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
    print(f"R-squared (R²): {r2:.4f}")

    return model, y_pred

# Initialize models
linear_model = LinearRegression()
random_forest_model = RandomForestRegressor(n_estimators=100, random_state=42)

# Train and evaluate models for Underpricing (%)
print("\n### Model Evaluation for Underpricing (%) ###")
lr_underpricing, lr_pred_up = train_and_evaluate_model(linear_model, X_train_up, X_test_up, y_train_up, y_test_up, "Underpricing (%)")
rf_underpricing, rf_pred_up = train_and_evaluate_model(random_forest_model, X_train_up, X_test_up, y_train_up, y_test_up, "Underpricing (%)")

# Train and evaluate models for One Year Return (%)
print("\n### Model Evaluation for One Year Return (%) ###")
lr_return, lr_pred_ret = train_and_evaluate_model(linear_model, X_train_ret, X_test_ret, y_train_ret, y_test_ret, "One Year Return (%)")
rf_return, rf_pred_ret = train_and_evaluate_model(random_forest_model, X_train_ret, X_test_ret, y_train_ret, y_test_ret, "One Year Return (%)")

print("\nModel training and evaluation completed.")
```

```python
# Cell 7: Model Interpretation & Feature Importance
# This cell analyzes feature importance to understand which factors influence IPO underpricing and returns.

import matplotlib.pyplot as plt
import numpy as np

# Function to plot feature importance
def plot_feature_importance(model, feature_names, target_name):
    if hasattr(model, "feature_importances_"):  # For tree-based models like RandomForest
        importance = model.feature_importances_
    elif hasattr(model, "coef_"):  # For linear models
        importance = np.abs(model.coef_)
    else:
        print(f"Feature importance not available for {model.__class__.__name__}.")
        return

    # Sort feature importance
    indices = np.argsort(importance)[::-1]
    sorted_features = [feature_names[i] for i in indices]
    sorted_importance = importance[indices]

    # Plot
    plt.figure(figsize=(10, 6))
    plt.barh(sorted_features, sorted_importance, color='skyblue')
    plt.xlabel("Feature Importance Score")
    plt.ylabel("Feature")
    plt.title(f"Feature Importance for {target_name} using {model.__class__.__name__}")
    plt.gca().invert_yaxis()
    plt.show()

# Extract feature names
feature_names = X.columns.tolist()

# Plot feature importance for Random Forest (Underpricing %)
print("\n### Feature Importance for Underpricing (%) ###")
plot_feature_importance(rf_underpricing, feature_names, "Underpricing (%)")

# Plot feature importance for Random Forest (One-Year Return %)
print("\n### Feature Importance for One-Year Return (%) ###")
plot_feature_importance(rf_return, feature_names, "One-Year Return (%)")

print("\nFeature importance analysis completed.")
```

```python
# Cell 8: Predictions & Final Insights
# This cell generates predictions using trained models and provides final insights.

import pandas as pd

# Generate predictions for test data
df_predictions = pd.DataFrame({
    "Actual Underpricing (%)": y_test_up.values,
    "Predicted Underpricing (LR)": lr_pred_up,
    "Predicted Underpricing (RF)": rf_pred_up,
    "Actual One Year Return (%)": y_test_ret.values,
    "Predicted One Year Return (LR)": lr_pred_ret,
    "Predicted One Year Return (RF)": rf_pred_ret
})

# Display sample predictions
print("\n### Sample Predictions ###")
display(df_predictions.head(10))

# Calculate and display the average error for each model
df_predictions["Error Underpricing (LR)"] = abs(df_predictions["Actual Underpricing (%)"] - df_predictions["Predicted Underpricing (LR)"])
df_predictions["Error Underpricing (RF)"] = abs(df_predictions["Actual Underpricing (%)"] - df_predictions["Predicted Underpricing (RF)"])
df_predictions["Error One Year Return (LR)"] = abs(df_predictions["Actual One Year Return (%)"] - df_predictions["Predicted One Year Return (LR)"])
df_predictions["Error One Year Return (RF)"] = abs(df_predictions["Actual One Year Return (%)"] - df_predictions["Predicted One Year Return (RF)"])

# Compute mean absolute error for each model
mae_results = {
    "Underpricing (LR)": df_predictions["Error Underpricing (LR)"].mean(),
    "Underpricing (RF)": df_predictions["Error Underpricing (RF)"].mean(),
    "One Year Return (LR)": df_predictions["Error One Year Return (LR)"].mean(),
    "One Year Return (RF)": df_predictions["Error One Year Return (RF)"].mean(),
}

# Display mean absolute error results
print("\n### Mean Absolute Errors ###")
for model, error in mae_results.items():
    print(f"{model}: {error:.4f}")

# Final Insights
print("\n### Final Insights ###")
print("- The model with the lowest MAE is likely the most reliable.")
print("- Random Forest generally performs better for non-linear relationships, while Linear Regression handles linear trends well.")
print("- Feature importance analysis (from Cell 7) can guide improvements in model performance.")

print("\nPredictions & final insights completed.")
```
